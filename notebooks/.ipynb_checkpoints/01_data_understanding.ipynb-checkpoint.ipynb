| Event Type | Number of Records |
|---|---:|

| Cart | 926,516 |

#### Observations

- **Views dominate the dataset**, with more than 40 million product-view events.
- Cart events are much less frequent than views, showing that only a relatively small proportion of browsing activity progresses to the cart stage.
- Purchase events are less frequent than cart events, which is expected in an e-commerce customer journey.
- These three event types allow the project to analyse a basic customer funnel:

`View → Cart → Purchase`

- No unexpected event types were found in the October dataset, so `event_type` appears consistent at this stage.

# STEP 1 — Data Understanding

## Dataset Overview

The project uses the **REES46 Multi-Category E-commerce Behaviour Dataset**. For the initial analysis, the October 2019 dataset (`2019-Oct.csv`) was selected.

The purpose of Day 1 was to understand the structure, scale, contents, and grain of the dataset before performing any cleaning or transformation.

## Dataset Size

- **Rows:** 42,448,764
- **Columns:** 9
- **Memory usage:** approximately 2.8+ GB

The dataset contains more than **42 million event-level records**, making it suitable for demonstrating large-scale data processing and cleaning using Python.

## Columns

The dataset contains the following fields:

- `event_time`
- `event_type`
- `product_id`
- `category_id`
- `category_code`
- `brand`
- `price`
- `user_id`
- `user_session`

## Data Types

| Column | Data Type |
|---|---|
| event_time | object |
| event_type | object |
| product_id | int64 |
| category_id | int64 |
| category_code | object |
| brand | object |
| price | float64 |
| user_id | int64 |
| user_session | object |

### Initial Data-Type Observations

- `event_time` is currently stored as an `object` rather than a datetime datatype and will require transformation before time-based analysis.
- `event_type`, `category_code`, `brand`, and `user_session` are stored as object columns and may contribute significantly to memory usage.
- `product_id`, `category_id`, and `user_id` are stored numerically but represent identifiers rather than numerical measurements.
- `price` is stored as `float64`.

## Date Coverage

The dataset covers the complete month of **October 2019**.

- **Earliest event:** 1 October 2019 at 00:00:00 UTC
- **Latest event:** 31 October 2019 at 23:59:59 UTC

This confirms that the dataset represents one complete calendar month of e-commerce activity.

### Observations

- Purchases occur less frequently than cart events.

- The event structure provides a basic customer journey:

## Dataset Grain

**One row represents one recorded user interaction with a specific product at a specific point in time.**

The interaction can represent:

- Viewing a product
- Adding a product to the cart
- Purchasing a product

Each event also contains associated information such as the user, product, category, brand, price, timestamp, and user session.

Understanding this grain is important because multiple rows belonging to the same user, product, or session are expected and should not automatically be considered duplicates.
